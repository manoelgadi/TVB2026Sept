```python
#client = OpenAI()
# This is an example of a chat completion
# We can use it to answer a question

response = myclient.chat.completions.create(
  model="gpt-4o",
  messages=[
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "Who won the last world series?"}
  ]
)
print(response.choices[0].message.content)
```

```python
# Let's ask another examle question:
question = "What are the most common cars?  I want a short list of 5 with 10 words per item"
response = myclient.chat.completions.create(
  model="gpt-3.5-turbo",
  messages=[
        {"role": "user", "content": question}
    ]
)
print(response.choices[0].message.content)
```

```python
# Let's time it so we know how long it takes:
import timeit
from timeit import default_timer as timer
start = timer()

question = "What are the most common cars?  I want a short list of 5 with 10 words per item"
response = myclient.chat.completions.create(
  model="gpt-3.5-turbo",
  messages=[
        {"role": "user", "content": question}
    ]
)

end = timer()
print(response.choices[0].message.content)
print('That took ',end - start,' seconds')
```

```python
# Let's try the same question with another model, ie Chat GPT model 4.0:
start = timer()
question = "What are the most common cars?  I want a short list of 5 with 10 words per item"
response = myclient.chat.completions.create(
  #model="gpt-4",
  model="gpt-4o",
  messages=[
        {"role": "user", "content": question}
    ]
)
end = timer()
print(response.choices[0].message.content)
print('That took ',end - start,' seconds')
```

```python
import openai
from pathlib import Path

# Define the output file path
speech_file_path = "./speech.mp3"

# Create the speech using OpenAI's latest API structure
response = myclient.audio.speech.create(
    model="tts-1",
    voice="alloy",
    input="Welcome to our Tech Venture Lab February 2025 program in the Tower in Madrid."
)

# Save the generated speech to a file
with open(speech_file_path, "wb") as audio_file:
    audio_file.write(response.content)

print(f"Speech saved to {speech_file_path}")

# Experiment with different voices: alloy, echo, fable, onyx, nova, shimmer
# These voices are optimized for English.
```

```python
# speech to text
# this will take that voice file and turn it back to text

#audio_file= open("/path/to/file/audio.mp3", "rb")
audio_file = open("speech.mp3", "rb")
transcript = myclient.audio.transcriptions.create(
  model="whisper-1",
  file=audio_file
)
print(transcript.text)
```

```python
# translate language from audio file
# this will transale a German audio file to English text

audio_file= open("german.mp3", "rb")
transcript = myclient.audio.translations.create(
  model="whisper-1",
  file=audio_file
)
print(transcript.text)
```

```python
# streaming
answer = myclient.chat.completions.create(
  model="gpt-4",
  messages=[
        {"role": "user", "content": 'write a poem of 100 words'}
    ],
    stream=True
)

for token in answer:
    print(token.choices[0].delta.content)
```
